```python
import cv2
import os
```

```python
# --- Configuration ---
output_dir = "yolo_frames_output"
os.makedirs(output_dir, exist_ok=True)
#interval_seconds = 2         # save one frame every N seconds
blur_threshold = 100         # higher = stricter blur filter
scene_diff_threshold = 30    # higher = stricter scene change detection
saved_count = 0
```

```python

def video2imagecapture(video_path, interval_seconds):
    global saved_count
    # --- Setup ---
    cap = cv2.VideoCapture(video_path)
    fps = int(cap.get(cv2.CAP_PROP_FPS))
    frame_count = 0
    
    while True:
        ret, frame = cap.read()
        if not ret:
            break #Exit the loop when video ends
        
        # Save at fixed interval
        if frame_count % (fps * interval_seconds) == 0:
            # Save frame
            frame_filename = os.path.join(output_dir, f"frame_{saved_count:04d}.jpg")
            cv2.imwrite(frame_filename, frame)
            saved_count += 1
            
        frame_count += 1

    cap.release()
    print(f"Extracted {saved_count} diverse frames to '{output_dir}'")
```

```python

# Supported video extensions
video_extensions = (".mp4", ".avi", ".mov", ".mkv")

def loadVideo(input_dir):
    for filename in os.listdir(input_dir):
        if filename.lower().endswith(video_extensions):
            video_path = os.path.join(input_dir, filename)
            print(f"Processing video: {video_path}")        
            
            video2imagecapture(video_path, 2)
#video2imagecapture(r"C:\Users\kamal\Downloads\14855642_2160_3840_30fps.mp4", 2)
#video2imagecapture(r"C:\Users\kamal\Downloads\153283-804933523_medium.mp4", 3)
#video2imagecapture(r"C:\Users\kamal\Downloads\15035729_2160_3840_30fps.mp4", 2)
```
